```python
from camp_project.markov import empirical_frequency, simulate, stationary_distribution
```

```python
P = [[0.8, 0.2], [0.3, 0.7],]
```

```python
T = 10_000

path = simulate(P, T)
```

```python
frequency = empirical_frequency(path)
stationary = stationary_distribution(P)

print("Empirical frequency:", frequency)
print("Stationary distribution:", stationary)
```

```python
running_frequency = []

count_ones = 0

for i, state in enumerate(path, start=1):
    if state == 1:
        count_ones += 1
    running_frequency.append(count_ones / i)
```

```python
len(running_frequency)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(running_frequency)

plt.axhline(
    stationary[1],
    linestyle="--",
    label="Stationary frequency",
)

plt.xlabel("Period")
plt.ylabel("Frequency of state 1")
plt.title("State 1 Frequency Over Time")
plt.legend()
plt.savefig("../paper/figures/convergence.pdf")
plt.show()
```
